```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd

%matplotlib inline
```

```python
file_path = r"C:\Users\ADITI\Documents\Ansys 2019 R1\1_0.33_ellipse_10deg_Re1000\fft_1000_cl"
data = pd.read_csv(file_path, skiprows= 4, header= None, sep= "\t")
data.columns = ['frequency', 'spectral_amplitude']
```

```python
freq_m = data["frequency"][data["spectral_amplitude"].idxmax()]
print(f"freq_m, {freq_m}")

# parameters
init_veloity = 0.001
Re = 1000

major_axis = 1
minor_axis = 0.33
aspect_ratio = 3
angle_of_attack = 10

strouhal_number = float(freq_m)*major_axis/init_veloity
print(f"strouhal_number, {strouhal_number}")
```

```python
# Read the existing dataset
# dataset = pd.DataFrame(columns = ['major_axis', 'minor_axis', 'Aspect_Ratio', 'init_velocity','Angle_of_attack','Reynolds_Number', 'freq_m', 'Strouhal_Number'])
dataset = pd.read_csv('dataset.txt', sep= '\t')
dataset
```

```python
# Adding the new data to the dataset
new_row = pd.DataFrame({'major_axis': [major_axis], 'minor_axis': [minor_axis], 'Aspect_Ratio': [aspect_ratio], 'init_velocity': [init_veloity], 'Angle_of_attack': [angle_of_attack],'Reynolds_Number': [Re], 'freq_m': [freq_m], 'Strouhal_Number': [strouhal_number]})
dataset = pd.concat([dataset, new_row], ignore_index=True)
dataset.to_csv('dataset.txt', sep= '\t', index= True)
dataset
```

```python
dataset.drop(columns= ['Unnamed: 0'], inplace= True)
# dataset.drop(0, inplace= True)
dataset.to_csv('dataset.txt', sep= '\t', index= True)
dataset
```
