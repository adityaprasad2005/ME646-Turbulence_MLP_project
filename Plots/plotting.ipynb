```python
# define libs

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

%matplotlib inline
```

```python
# load data

data = pd.read_csv('Dataset/Dataset.txt', sep= '\t')
data
```

```python
# plotting the 3D plots for a particular aspect ratio
from mpl_toolkits.mplot3d import Axes3D

aspect_ratio = 2

data_AR = data[data["Aspect_Ratio"]== aspect_ratio]
Res = data_AR["Reynolds_Number"]
Strouhals = data_AR["Strouhal_Number"]
AoAs = data_AR["Angle_of_attack"]

fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

ax.scatter(AoAs, Res, Strouhals, c=AoAs, cmap='viridis')
ax.set_xlabel("Aspect Ratio")
ax.set_ylabel("Reynolds Number")
ax.set_zlabel("Strouhal Number")
```

```python
# plot 2D plots to show the trendline

aspect_ratio = 2

line_styples = ["-", "--",  ":"]
colors= ["b", "g", "r"]
markers = ["o", "v", "^"]
alphas = [0.5, 0.6, 0.7]
offset = 0.005

data_AR = data[data["Aspect_Ratio"]== aspect_ratio]

unique_Re = data_AR["Reynolds_Number"].unique()
i = 0
for Re in unique_Re:
    data_Re = data_AR[data_AR["Reynolds_Number"] == Re]
    plt.plot(data_Re["Angle_of_attack"], data_Re["Strouhal_Number"]+ i*offset, marker=markers.pop() , color = colors.pop(),label="Re = " + str(Re), alpha= alphas.pop())
    i += 1
plt.xlabel("Angle of Attack")
plt.ylabel("Strouhal Number")
plt.legend()
plt.title("Elliptic Cylinder with Aspect Ratio = " + str(aspect_ratio))
plt.savefig("Strouhal_vs_AoA.png")
plt.show()
```

```python
aspect_ratio = 2

line_styples = ["-", "--",  ":"]
colors= ["b", "g", "r"]
markers = ["o", "v", "^"]
alphas = [0.5, 0.8, 0.9]

data_AR = data[data["Aspect_Ratio"]== aspect_ratio]

unique_AoA = data_AR["Angle_of_attack"].unique()
for AoA in unique_AoA:
    data_AoA = data_AR[data_AR["Angle_of_attack"] == AoA]
    plt.plot(data_AoA["Reynolds_Number"], data_AoA["Strouhal_Number"], color = colors.pop(), linestyle=line_styples.pop(), marker=markers.pop() , alpha= alphas.pop() ,label="AoA = " + str(AoA))

plt.xlabel("Reynolds Number")
plt.ylabel("Strouhal Number")
plt.xticks(data_AR["Reynolds_Number"].unique())
plt.legend()
plt.title("Elliptic Cylinder with Aspect Ratio = " + str(aspect_ratio))
plt.savefig("Strouhal_vs_Re.png")
plt.show()
```
